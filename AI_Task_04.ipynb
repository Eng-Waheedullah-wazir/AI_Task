```python
print("Name:Waheed Ullah")
print("Roll Number:23p-3079")

class SimpleReflex:
    def __init__(self):
        self.signalrule ={ 
        "red":"Stop",
        "yellow":"slow down",
        "green":"Go"
        }

    def sensor(self,signal):
        
        if signal == 'red' or signal == "Red":
            return 'red'

        elif signal == "yellow" or signal == "Yellow":
            return "yellow"

        elif signal == "green"  or signal == "Green":
            return 'green'
        else :
            return "dn"

    def act(self,percept):
        return self.signalrule.get(percept, "dn");

    def run(self,signaltype):
        input1=self.sensor(signaltype)
        action=self.act(input1)
        print(f'Signal:{signaltype} | percept is {input1}  Action is :{action}')
        

        





object1=SimpleReflex()
signal=["green",'red','yellow','green','red']
for i in signal:
    object1.run(i)
```
