```python
print("Waheed Ullah")
print("23p-3079")

student_database = []
list1_of_tuple = []

student_database.insert(0, ("Farhad", 22, 85))
student_database.insert(1, ("Amir", 23, 70))
student_database.insert(2, ("marwan", 18, 90))

print("Show the Total Number of the student:")
print(len(student_database))

for i in range(len(student_database)):
    print(student_database[i])

for i in range(len(student_database)):
    list1_of_tuple.append(list(student_database[i]))

maximum = list1_of_tuple[0]
for i in range(len(list1_of_tuple)):
    if list1_of_tuple[i][2] > maximum[2]:
        maximum = list1_of_tuple[i]

print("The student with highest grade")
print(maximum)

minium_age = list1_of_tuple[0]
for i in range(len(list1_of_tuple)):
    if list1_of_tuple[i][1] < minium_age[1]:
        minium_age = list1_of_tuple[i]

list1_of_tuple.remove(minium_age)
print("After removing the lowest age student:")
print(list1_of_tuple)

sortage = []
temp = list(student_database)

while temp:
    young = temp[0]
    for student in temp:
        if student[1] < young[1]:
            young = student
    sortage.append(young)
    temp.remove(young)

print(sortage)
```

```python
print("Waheed Ullah")
print("23p-3079")

smartchatbot = {
    "greeting": {
        "keyword": ("hi", "hello", "hey"),
        "response": ["Hi!", "How can I help you?"]
    },
    "help": {
        "keyword": ("help", "support", "assist"),
        "response": ["I am here to help you."]
    },
    "exit": {
        "keyword": ("exit", "quit", "bye"),
        "response": ["Goodbye!"]
    }
}

history = []
defaultresponse = "Sorry, I don't understand that."

while True:
    userinput = input("You: ").lower()
    history.append(userinput)

    matched = False

    for category in smartchatbot:
        keywords = smartchatbot[category]["keyword"]

        if any(word in userinput for word in keywords):
            print("Bot:", smartchatbot[category]["response"][0])
            matched = True

            if category == "exit":
                print("User Input History:", history)
                exit()

            break

    if not matched:
        print("Bot:", defaultresponse)
```
