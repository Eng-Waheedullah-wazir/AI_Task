```python
print("NAME: Waheed Ullah")
print("ROLL NO : 23P-3079")

import pandas as pd
import numpy as np

data = {
    "Employee ID": [1, 2, 3, 4, 5, 2, None, 8],
    "Name": ["Ali Khan", "Sara Ali", "John Smith", "Maryam Noor", None, "Sara Ali", "David Lee", "Emma Khan"],
    "Department": ["IT", "I.T.", "HR", "Information Technology", "Sales", "I.T.", "HR", "IT"],
    "Salary": ["50000", None, "60000", None, "45000", None, "52000", "58000"],
    "Join Date": ["2020-01-01", "2019-05-10", "2021-03-15", "2018-07-22", "2022-02-01", "2019-05-10", "2020-11-11", "2017-09-09"],
    "Email": ["ali.khan@company.com", None, "john.smith@company.com", None, "maryam.noor@company.com", None, "david.lee@company.com", None]
}

df = pd.DataFrame(data)

print("Before Cleaning:")
print(df)
print("\nMissing values:")
print(df.isnull().sum())

df = df.drop_duplicates()
df = df.dropna(subset=["Employee ID", "Name"])

df["Department"] = df["Department"].replace({
    "I.T.": "IT",
    "Information Technology": "IT"
})

df["Salary"] = pd.to_numeric(df["Salary"], errors='coerce')

df["Salary"] = df.groupby("Department")["Salary"].transform(
    lambda x: x.fillna(x.median())
)

for index, row in df.iterrows():
    if pd.isnull(row["Email"]):
        first, last = row["Name"].lower().split()
        df.at[index, "Email"] = f"{first}.{last}@company.com"

df["Join Date"] = pd.to_datetime(df["Join Date"])

print("\nAfter Cleaning:")
print(df)







```

```python
print("NAME: Waheed Ullah")
print("ROLL NO : 23P-3079")
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()
edges = [
    ("A", "B", 2), ("A", "C", 4),
    ("B", "D", 3), ("B", "E", 5),
    ("C", "D", 1), ("C", "F", 6),
    ("D", "E", 2), ("D", "F", 3),
    ("E", "F", 1)
]

G.add_weighted_edges_from(edges)
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True)
labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=labels)
plt.show()
path = nx.shortest_path(G, "A", "F", weight='weight')
time = nx.shortest_path_length(G, "A", "F", weight='weight')

print("Shortest Path from A to F:", path)
print("Total Travel Time:", time)
centrality = nx.betweenness_centrality(G)
print("Betweenness Centrality:", centrality)

most_central = max(centrality, key=centrality.get)
print("Most Central City:", most_central)
```

```python
print("NAME: Waheed Ullah")
print("ROLL NO : 23P-3079")

import networkx as nx
import matplotlib.pyplot as plt

DG = nx.DiGraph()

dependencies = [
    (1,2),(1,3),
    (2,4),(2,5),
    (3,5),
    (4,6),
    (5,6),(5,7),
    (6,8),
    (7,8)
]

DG.add_edges_from(dependencies)
pos = nx.spring_layout(DG)
nx.draw(DG, pos, with_labels=True)
plt.show()
order = list(nx.topological_sort(DG))
print("Valid Task Order:", order)
print("Is DAG:", nx.is_directed_acyclic_graph(DG))
critical_path = nx.dag_longest_path(DG)
print("Critical Path:", critical_path)



```
