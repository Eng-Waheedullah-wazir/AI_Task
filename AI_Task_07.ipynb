**TASK 1**

**8 Queens Problem using CSP**


```python
print("Name:Waheed Ullah")
print("Roll NO:23p-3079")

N = 8
board = [-1] * N
def is_safe(row, col):
    for i in range(row):
        if board[i] == col:
            return False
        if abs(board[i] - col) == abs(i - row):
            return False
    return True
def solve(row):
    if row == N:
        return True
    for col in range(N):
        if is_safe(row, col):
            board[row] = col
            if solve(row + 1):
                return True
            board[row] = -1
    return False
if solve(0):
    print("\n8-Queens Solution:\n")
    for i in range(N):
        for j in range(N):
            if board[i] == j:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print("\nQueen Positions:\n")
    for i in range(N):
        print(f"Row {i + 1} -> Column {board[i] + 1}")
else:
    print("No Solution Found")
```

**TASK 2 A**


**BASE+BALL=GAME**

```python
print("Name:Waheed Ullah")
print("Roll NO:23p-3079")
from itertools import permutations
letters = ('B', 'A', 'S', 'E', 'L', 'G', 'M')
digits = range(10)
found = False
for perm in permutations(digits, len(letters)):
    B, A, S, E, L, G, M = perm
    if B == 0 or G == 0:
        continue
    BASE = 1000 * B + 100 * A + 10 * S + E
    BALL = 1000 * B + 100 * A + 10 * L + L
    GAMES = 10000 * G + 1000 * A + 100 * M + 10 * E + S
    if BASE + BALL == GAMES:
        found = True
        print("\nSolution Found:\n")
        print("Digit Mapping:")
        print(f"B = {B}")
        print(f"A = {A}")
        print(f"S = {S}")
        print(f"E = {E}")
        print(f"L = {L}")
        print(f"G = {G}")
        print(f"M = {M}")
        print("\nVerification:")
        print(f"  {BASE}")
        print(f"+ {BALL}")
        print("-------")
        print(f"{GAMES}")
        break

if not found:
    print("No Solution Found")
```

**TASK 2 B**


**Your+You =Heart**

```python
print("Name:Waheed Ullah")
print("Roll NO:23p-3079")

from itertools import permutations
letters = ('Y', 'O', 'U', 'R', 'H', 'E', 'A', 'T')
digits = range(10)
found = False
for perm in permutations(digits, len(letters)):
    Y, O, U, R, H, E, A, T = perm
    if Y == 0 or H == 0:
        continue
    YOUR = 1000 * Y + 100 * O + 10 * U + R
    YOU = 100 * Y + 10 * O + U
    HEART = 10000 * H + 1000 * E + 100 * A + 10 * R + T
    if YOUR + YOU == HEART:
        found = True
        print("\nSolution Found:\n")
        print("Digit Mapping:")
        print(f"Y = {Y}")
        print(f"O = {O}")
        print(f"U = {U}")
        print(f"R = {R}")
        print(f"H = {H}")
        print(f"E = {E}")
        print(f"A = {A}")
        print(f"T = {T}")

        print("\nVerification:")
        print(f"  {YOUR}")
        print(f"+  {YOU}")
        print("--------")
        print(f"{HEART}")
        break
if not found:
    print("No Solution Found")
```

**TASK 3 — Map Coloring**

```python
print("Name:Waheed Ullah")
print("Roll NO:23p-3079")

colors = ["Red", "Green", "Blue"]
regions = ["A", "B", "C", "D", "E"]
graph = {
    "A": ["B", "C"],
    "B": ["A", "C", "D"],
    "C": ["A", "B", "D", "E"],
    "D": ["B", "C", "E"],
    "E": ["C", "D"]
}
assignment = {}
def is_safe(region, color):
    for neighbor in graph[region]:
        if neighbor in assignment and assignment[neighbor] == color:
            return False
    return True

def solve(index):
    if index == len(regions):
        return True
    region = regions[index]
    for color in colors:
        if is_safe(region, color):
            assignment[region] = color
            if solve(index + 1):
                return True
            del assignment[region]
    return False

if solve(0):
    print("\nMap Coloring Solution:\n")
    for region in regions:
        print(f"{region} -> {assignment[region]}")
else:
    print("No Solution Found")
```
